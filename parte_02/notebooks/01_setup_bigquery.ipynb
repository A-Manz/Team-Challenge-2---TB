# Setup BigQuery — NDIVIA
Conexión a Google BigQuery y creación del dataset `ndivia`.

```python
from google.cloud import bigquery
from google.oauth2 import service_account
import os
from dotenv import load_dotenv

load_dotenv()

# Configuración
PROJECT_ID = os.getenv("GCP_PROJECT_ID")
DATASET_ID = os.getenv("BQ_DATASET_ID")
CREDENTIALS_PATH = os.getenv("GOOGLE_APPLICATION_CREDENTIALS")

# Cliente autenticado
credentials = service_account.Credentials.from_service_account_file(
    CREDENTIALS_PATH
)
client = bigquery.Client(
    project=PROJECT_ID,
    credentials=credentials
)

# Crear dataset
dataset_ref = bigquery.Dataset(f"{PROJECT_ID}.{DATASET_ID}")
dataset_ref.location = "EU"  # Datos en Europa
dataset = client.create_dataset(dataset_ref, exists_ok=True)
print(f"Dataset {DATASET_ID} creado")
```

```python
# CUSTOMERS

table_customers = f"{PROJECT_ID}.{DATASET_ID}.customers"

schema = [
    bigquery.SchemaField("id", "INT64"),
    bigquery.SchemaField("full_name", "STRING"),
    bigquery.SchemaField("contact_email", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("register_date", "TIMESTAMP"),
    bigquery.SchemaField("country", "STRING"),
    bigquery.SchemaField("city", "STRING"),
    bigquery.SchemaField("discovery_source", "STRING"),
]

table = bigquery.Table(table_customers, schema=schema)
table = client.create_table(table, exists_ok=True)
print("Creada la tabla {}.{}.{}".format(table.project, table.dataset_id, table.table_id))
```

```python
# CATEGORIES

table_categories = f"{PROJECT_ID}.{DATASET_ID}.categories"

schema = [
    bigquery.SchemaField("cat_id", "INT64"),
    bigquery.SchemaField("category", "STRING"),
    bigquery.SchemaField("description", "STRING"),
]

table = bigquery.Table(table_categories, schema=schema)
table = client.create_table(table, exists_ok=True)
print("Creada la tabla {}.{}.{}".format(table.project, table.dataset_id, table.table_id))
```

```python
# PRODUCTS

table_products = f"{PROJECT_ID}.{DATASET_ID}.products"

schema = [
    bigquery.SchemaField("product_id", "INT64"),
    bigquery.SchemaField("product_name", "STRING"),
    bigquery.SchemaField("cat_id", "INT64"),
    bigquery.SchemaField("retail_price", "FLOAT64"),
    bigquery.SchemaField("vendor_price", "FLOAT64"),
    bigquery.SchemaField("stock", "INT64"),
    bigquery.SchemaField("active", "BOOL", mode="REQUIRED"),
]

table = bigquery.Table(table_products, schema=schema)
table = client.create_table(table, exists_ok=True)
print("Creada la tabla {}.{}.{}".format(table.project, table.dataset_id, table.table_id))
```

```python
# ORDERS

table_orders = f"{PROJECT_ID}.{DATASET_ID}.orders"

schema = [
    bigquery.SchemaField("order_id", "INT64"),
    bigquery.SchemaField("status", "STRING"),
    bigquery.SchemaField("customer_id", "INT64"),
    bigquery.SchemaField("info", "STRING"),
    bigquery.SchemaField("purchase_date", "TIMESTAMP"),
    bigquery.SchemaField("dispatch_date", "TIMESTAMP"),
    bigquery.SchemaField("delivery_date", "TIMESTAMP"),
]

table = bigquery.Table(table_orders, schema=schema)
table = client.create_table(table, exists_ok=True)
print("Creada la tabla {}.{}.{}".format(table.project, table.dataset_id, table.table_id))
```
